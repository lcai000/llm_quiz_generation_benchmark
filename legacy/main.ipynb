```python
from prompts import Math
quiz = CreateQuiz()
math_prompt = Math()
math_prompt.quantity = 10
multiplication_prompt = math_prompt.algebraic_equations()
user_prompt, system_prompt = multiplication_prompt["user_prompt"],multiplication_prompt["system_prompt"]
multiplication_quiz = quiz(batch_size=10,
                            batches=3,
                            prompt=user_prompt,
                            system_prompt=system_prompt,
                            mode="gemini")
print(multiplication_quiz)
with open("example_result.txt","w") as f:
    f.write(multiplication_quiz["MODEL"])
    f.write("\n")
    f.write(str(multiplication_quiz["QUANTITY"]))
    f.write("\n")
    f.write(str(multiplication_quiz["BATCH_SIZE"]))
    f.write("\n")
    f.write(multiplication_quiz["QUIZ"])
    f.write("\n")
```
